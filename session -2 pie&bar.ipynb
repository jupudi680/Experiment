```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
pd.DataFrame(zip(keys,values),columns=['class','frequency'])
```

**Bar chart**
- X-axis:class names(keys)
- Y-axis:frequency(values)

```python
plt.bar(keys,values)
```

```python
colors=['Blue','pink','red','purple','green','violet']
plt.figure(figsize=(10,4))
plt.bar(keys,values,color=colors)
plt.xlabel('continent')
plt.ylabel('No of Applicents')
plt.title("Bar chart")
plt.savefig('continent_bar.jpg')
plt.show()
```

**Pie chart**

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        explode=[0.01 for i in range(6)],
        labels=keys,
        radius=0.8,
        autopct='%1.2f%%',
        startangle=40)
plt.savefig('continent_pie.jpg')
plt.show()
```

**pie chart**
- Autopct=%a.bf%%
- first % is the formatting
- a : how much min width
- b : after point how many decimals
- %% : % symbol
- %1.1f%% : 40.2%
- %2.2f%% : 423.23
