**KNNImputer**

- KNN : K-nearest neighbours
- where k is a number
- insted of taking avg of all the values data
    - first will find the neighbours who are close with missing data
    - by using euclidian distance calculation will identify the neighbours rows
    - suppose k=2 will identify two closeset rows
    - those closest row values will take it will calculate the mean
    - finally will fill with the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),columns=['Salary','Age'])
df
```

```python
from sklearn.impute import KNNImputer
knn=KNNImputer(n_neighbors=2)
pd.DataFrame(knn.fit_transform(df))
```
